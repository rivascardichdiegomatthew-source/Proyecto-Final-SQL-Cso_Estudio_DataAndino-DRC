17- ¿Los hijos aseguran el viaje? Escribe una consulta usando CASE WHEN que clasifique a la reserva como "Con Hijos" (si children > 0 o babies > 0) o "Sin Hijos". Agrupa por esta nueva categoría y calcula la tasa de cancelación (puedes sacar el promedio - AVG - de la columna is_canceled). ¿Qué segmento cancela más?


```python
%sql
SELECT
  CASE WHEN children > 0 OR babies > 0 THEN 'Con Hijos' ELSE 'Sin Hijos' END AS segmento,
  COUNT(*)                       AS total_reservas,
  ROUND(AVG(is_canceled), 4)     AS tasa_cancelacion
FROM cursodataacademy.hotel_db.tb_reservas
GROUP BY segmento
ORDER BY tasa_cancelacion DESC;
```

Conclusión: Sin Hijos cancela más (37.2% frente a 34.9%), así que tener hijos asegura el viaje solo en parte. La diferencia es de apenas 2.3 puntos, un efecto pequeño que no justifica por sí solo una política de contratos.

19- ¿Cambiar la reserva es señal de compromiso? Al igual que en los ejercicios anteriores, utiliza CASE WHEN para segmentar la columna booking_changes en tres grupos: "Sin Cambios" (0), "1 a 3 cambios" y "Más de 3 cambios". Calcula la tasa de cancelación para cada grupo. ¿Quiénes tienen menor riesgo de cancelar?


```python
%sql
SELECT
  CASE
    WHEN booking_changes = 0 THEN 'Sin Cambios'
    WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
    ELSE 'Más de 3 cambios'
  END AS grupo_cambios,
  COUNT(*)                   AS total_reservas,
  ROUND(AVG(is_canceled), 4) AS tasa_cancelacion
FROM cursodataacademy.hotel_db.tb_reservas
GROUP BY grupo_cambios
ORDER BY tasa_cancelacion ASC;
```

Conclusión: Quienes hacen de 1 a 3 cambios tienen el menor riesgo de cancelar (15.6%), muy por debajo de quienes no cambian nada (40.9%). Modificar la reserva sí es señal de compromiso. El grupo de más de 3 cambios (18.0%) tiene solo 643 reservas, por lo que esa cifra es menos confiable.
